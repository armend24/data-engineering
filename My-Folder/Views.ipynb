```python
%sql
insert into test
values (1, 'Armend'),(2, 'Brenda'), (3, 'Laforteza'), (4, 'Mariano'), (5, 'Ricardo'), (6, 'Sophia'), (7, 'Moonlight');
```

```python
%sql
-- This is a stored view, available across Spark and cluster sessions. It is essentially a stored query.
create view test_view as
select name,
case
    when name ilike '%a%' then true
    else false
end as has_a
from test
```

```python
%sql
select * from test_view;
```

```python
%sql
-- This temporary view is available only within the current Spark session, so it cannot be accessed outside this notebook. If the notebook is detached, the Spark session is gone, and with it the temporary view below.
create temporary view temp_test_view as
select *, 
case
    when name ilike '%o%' then true
    else false
end as contains_o
from test;
```

```python
%sql
select * from temp_test_view
```

```python
%sql
-- This global temporary view is available across notebooks and Spark sessions on the same cluster. It is scoped to the cluster's shared Spark application, so it remains available until the cluster is restarted or terminated. When querying it, you have to call it from the 'global_temp' database: SELECT * FROM global_temp.global_temp_test_view
create global temporary view global_temp_test_view as
select *, 
case
    when name ilike '%o%' then true
    else false
end as contains_o
from test;
```

```python
%sql
select * from global_temp.global_temp_test_view
```
