```python
%sql
MERGE INTO books.csv_books b
USING (
    SELECT * FROM read_files(
        '${dataset.bookstore}/books-csv-new',
        format => 'csv',
        header => true,
        delimiter => ';'
    )
) nb
ON b.book_id = nb.book_id
WHEN MATCHED AND b.title IS NULL
    THEN UPDATE SET b.title = nb.title
WHEN NOT MATCHED
    THEN INSERT *
```
