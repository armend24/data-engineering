```python
%sql
-- Temporary views exist only in the current Spark session, so querying one from another notebook/session will fail.
select * from temp_test_view;
-- If the cluster shuts down or the notebook is detached, this view is gone.
```

```python
%sql
-- Global temporary views are shared across notebooks attached to the same Spark cluster, so this query should succeed.
select * from global_temp.global_temp_test_view;
-- If the cluster shuts down, both temporary and global temporary views are removed.
```
