```python
%sql
-- first we put the data we want to upsert into the 'ordesr' table
create or replace temporary view customer_updates as
select * from json.`${dataset.bookstore}/customers-json-new`;

MERGE INTO customers c
USING customer_updates cu
ON c.customer_id = cu.customer_id
WHEN MATCHED AND c.email IS NULL AND cu.email IS NOT NULL THEN
    UPDATE SET c.email = cu.email, updated = cu.updated
WHEN NOT MATCHED THEN INSERT *
    
```
